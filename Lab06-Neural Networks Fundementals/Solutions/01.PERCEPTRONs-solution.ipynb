### What is a Perceptron?

A perceptron is the simplest neural network unit. It:
1. Takes **binary inputs** (0 or 1)
2. Multiplies each input by a **weight**
3. Sums the results
4. Compares to a **threshold** to make a decision

**Formula:**
$$\text{output} = \begin{cases} 1 & \text{if } \sum_{i} x_i \cdot w_i \geq \theta \\ 0 & \text{otherwise} \end{cases}$$

Where $x_i$ = inputs, $w_i$ = weights, $\theta$ = threshold.

## The Beach Decision Problem

Should we go to the beach? We have 6 criteria, each with a weight reflecting its importance:

| # | Criterion | Weight |
|---|-----------|--------|
| 1 | Fun in the sun | +0.8 |
| 2 | Friends going | +0.7 |
| 3 | Money (cost / expense) | −0.5 |
| 4 | Transportation annoyance | −0.3 |
| 5 | Opportunity cost | −0.2 |
| 6 | Chance of whale watching | 0.0 |

**Threshold:** θ = 0.6

- Positive weights = reasons **to go**
- Negative weights = reasons **not to go**
- Input = 1 if the condition is true, 0 if false

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the weights and threshold
criteria = [
    "Fun in the sun",
    "Friends going",
    "Money (cost)",
    "Transportation annoyance",
    "Opportunity cost",
    "Whale watching chance"
]
weights = np.array([0.8, 0.7, -0.5, -0.3, -0.2, 0.0])
threshold = 0.6

print("Perceptron weights:")
for c, w in zip(criteria, weights):
    print(f"  {c:<30} w = {w:+.1f}")
print(f"\nDecision threshold: θ = {threshold}")
```

## Scenario A — Great Beach Day

It's a sunny day, your friends are all going, you have enough money, transportation is easy, no other commitments, and there's a small chance of seeing whales.

| Criterion | Input (x) | Weight (w) | x × w |
|-----------|-----------|------------|-------|
| Fun in the sun | 1 | +0.8 | +0.8 |
| Friends going | 1 | +0.7 | +0.7 |
| Money (cost) | 0 | −0.5 | 0.0 |
| Transportation | 0 | −0.3 | 0.0 |
| Opportunity cost | 0 | −0.2 | 0.0 |
| Whale watching | 1 | 0.0 | 0.0 |

**Weighted sum = 0.8 + 0.7 + 0 + 0 + 0 + 0 = 1.5**

**Decision: 1.5 ≥ 0.6 → GO TO THE BEACH! ✅**

```python
def perceptron_decision(inputs, weights, threshold, scenario_name):
    """Compute perceptron output with step-by-step breakdown."""
    print(f"--- {scenario_name} ---")
    total = 0
    for c, x, w in zip(criteria, inputs, weights):
        contribution = x * w
        total += contribution
        symbol = "✓" if x == 1 else " "
        print(f"  [{symbol}] {c:<30}  x={x}  w={w:+.1f}  contribution={contribution:+.2f}")
    print(f"\n  Weighted sum = {total:.2f}")
    print(f"  Threshold    = {threshold}")
    if total >= threshold:
        print("  Decision: GO TO THE BEACH! 🏖️")
    else:
        print("  Decision: Stay home. 🏠")
    print()
    return total

# Scenario A: great day
inputs_A = np.array([1, 1, 0, 0, 0, 1])  # sunny, friends going, no cost problem, easy transport, free, whale chance
sum_A = perceptron_decision(inputs_A, weights, threshold, "Scenario A: Great Beach Day")
```

## Scenario B — Not a Great Day

It's cloudy (no sun fun), your friends aren't going, it's expensive, transport is a hassle, and you have other things to do.

| Criterion | Input (x) | Weight (w) | x × w |
|-----------|-----------|------------|-------|
| Fun in the sun | 0 | +0.8 | 0.0 |
| Friends going | 0 | +0.7 | 0.0 |
| Money (cost) | 1 | −0.5 | −0.5 |
| Transportation | 1 | −0.3 | −0.3 |
| Opportunity cost | 1 | −0.2 | −0.2 |
| Whale watching | 0 | 0.0 | 0.0 |

**Weighted sum = 0 + 0 − 0.5 − 0.3 − 0.2 + 0 = −1.0**

**Decision: −1.0 < 0.6 → Stay home. 🏠**

```python
# Scenario B: bad conditions
inputs_B = np.array([0, 0, 1, 1, 1, 0])  # no sun, no friends, expensive, annoying transport, other plans
sum_B = perceptron_decision(inputs_B, weights, threshold, "Scenario B: Bad Beach Day")
```

## Scenario C — Mixed Day (Edge Case)

Sunny and friends are going, but it's expensive and transport is annoying.

```python
# Scenario C: mixed
inputs_C = np.array([1, 1, 1, 1, 0, 0])  # sunny, friends, expensive, bad transport
sum_C = perceptron_decision(inputs_C, weights, threshold, "Scenario C: Mixed Day")

print(f"Weighted sum for C = {sum_C:.2f}")
print("The positives (sun + friends = +1.5) just outweigh the negatives (money + transport = -0.8)")
print(f"Net = +0.7, which is >= threshold {threshold} → Go!")
```

## Visualizing the Decisions

```python
scenarios = ["A: Great Day", "B: Bad Day", "C: Mixed Day"]
sums = [sum_A, sum_B, sum_C]
colors = ['green' if s >= threshold else 'red' for s in sums]

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar(scenarios, sums, color=colors, edgecolor='black', alpha=0.8)
ax.axhline(y=threshold, color='blue', linestyle='--', linewidth=2, label=f'Threshold = {threshold}')
ax.axhline(y=0, color='black', linewidth=0.5)

for bar, s in zip(bars, sums):
    ax.text(bar.get_x() + bar.get_width()/2, s + 0.05, f'{s:.2f}',
            ha='center', va='bottom', fontweight='bold')

ax.set_ylabel('Weighted Sum')
ax.set_title('Perceptron Decisions: Go to the Beach?')
ax.legend()
ax.set_ylim(-1.5, 2.0)
plt.tight_layout()
plt.show()

print("Green bars = GO (weighted sum >= threshold)")
print("Red bars = STAY HOME (weighted sum < threshold)")
```

## Understanding the Weights

Notice how the weights encode "importance":
- **Fun in the sun (+0.8)** is the strongest positive factor
- **Friends going (+0.7)** is the second strongest positive
- **Whale watching (0.0)** has no influence on the decision
- Negative weights represent costs/obstacles

**Key insight:** A perceptron computes a *weighted vote*. Each input votes for or against the decision with a strength determined by its weight.

## Design Your Own Decision Problem

Use the template below to create your own perceptron for a different decision.

```python
# --- YOUR OWN DECISION PROBLEM ---
# Example: Should I study tonight?

my_criteria = [
    "Exam tomorrow",
    "Behind on material",
    "Tired",
    "Friends want to hang out",
    "Already studied today"
]
my_weights = np.array([0.9, 0.7, -0.4, -0.3, -0.2])
my_threshold = 0.5

# Current scenario
my_inputs = np.array([1, 1, 0, 1, 0])  # exam tomorrow, behind, not tired, friends calling, haven't studied yet

my_sum = np.dot(my_inputs, my_weights)
print("Decision: Should I study tonight?")
print(f"Weighted sum = {my_sum:.2f}, Threshold = {my_threshold}")
if my_sum >= my_threshold:
    print("YES — Study! 📚")
else:
    print("Take a break! 😴")
```

## Key Takeaways

1. **A perceptron makes binary decisions** by computing a weighted sum and comparing to a threshold
2. **Positive weights** push toward YES; **negative weights** push toward NO
3. **The threshold** controls how easy it is to trigger a YES
4. **This is the building block** for all neural networks — stacking perceptrons creates deeper networks
5. **Limitations:** Perceptrons can only separate linearly separable problems (can't solve XOR)
